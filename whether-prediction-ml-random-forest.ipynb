# Why try Random Forest?
Random Forest = many decision trees

Instead of using just one tree:

Tree 1 → Yes
Tree 2 → Yes
Tree 3 → No
Tree 4 → Yes
Tree 5 → Yes


The Random Forest looks at the predictions from many trees and uses their combined result.

Think of it like asking 100 people:

"Do you think it will rain tomorrow?"

If 80 say Yes and 20 say No, the forest predicts Yes.

That's the basic idea.
Our Logistic Regression assumes a fairly simple mathematical relationship between the features and the answer.

For example, rain might depend on combinations such as:

High humidity
      +
Low pressure
      +
Certain wind direction
      +
Rainfall today
      ↓
Higher chance of rain tomorrow

Random Forest can learn these kinds of non-linear relationships and feature combinations.

# One common technique is to tell the model:"Pay more attention when you get a Rain example wrong."


# What is class_weight="balanced"?
Now imagine the teacher says:

# "The Yes questions are more important. If you get one wrong, I'll penalize you more."
So when the model is learning, mistakes involving the minority class (Yes) receive more emphasis.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

thresholds = [0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60]

results = []

for threshold in thresholds:
    y_pred_threshold = np.where(y_prob >= threshold, "Yes", "No")
    
    accuracy = accuracy_score(y_test, y_pred_threshold)
    precision = precision_score(y_test, y_pred_threshold, pos_label="Yes")
    recall = recall_score(y_test, y_pred_threshold, pos_label="Yes")
    f1 = f1_score(y_test, y_pred_threshold, pos_label="Yes")
    
    results.append([
        threshold,
        accuracy,
        precision,
        recall,
        f1
    ])
```

```python
results_df = pd.DataFrame(
    results,
    columns=[
        "Threshold",
        "Accuracy",
        "Precision",
        "Recall",
        "F1"
    ]
)

results_df
```

```python
def predict_rain(
    location,
    min_temp,
    max_temp,
    rainfall,
    wind_gust_dir,
    wind_gust_speed,
    wind_dir_9am,
    wind_dir_3pm,
    wind_speed_9am,
    wind_speed_3pm,
    humidity_9am,
    humidity_3pm,
    pressure_9am,
    pressure_3pm,
    temp_9am,
    temp_3pm,
    rain_today,
    year,
    month,
    day
):
    
    # Create an empty row with all 109 features
    input_data = pd.DataFrame(
        0,
        index=[0],
        columns=loaded_features
    )
    
    # Numerical features
    input_data["MinTemp"] = min_temp
    input_data["MaxTemp"] = max_temp
    input_data["Rainfall"] = rainfall
    input_data["WindGustSpeed"] = wind_gust_speed
    input_data["WindSpeed9am"] = wind_speed_9am
    input_data["WindSpeed3pm"] = wind_speed_3pm
    input_data["Humidity9am"] = humidity_9am
    input_data["Humidity3pm"] = humidity_3pm
    input_data["Pressure9am"] = pressure_9am
    input_data["Pressure3pm"] = pressure_3pm
    input_data["Temp9am"] = temp_9am
    input_data["Temp3pm"] = temp_3pm
    input_data["RainToday"] = rain_today
    input_data["Year"] = year
    input_data["Month"] = month
    input_data["Day"] = day

    # One-hot encoded categorical features
    location_column = "Location_" + location
    gust_column = "WindGustDir_" + wind_gust_dir
    dir9_column = "WindDir9am_" + wind_dir_9am
    dir3_column = "WindDir3pm_" + wind_dir_3pm

    # Set the appropriate categories to 1
    if location_column in input_data.columns:
        input_data[location_column] = 1

    if gust_column in input_data.columns:
        input_data[gust_column] = 1

    if dir9_column in input_data.columns:
        input_data[dir9_column] = 1

    if dir3_column in input_data.columns:
        input_data[dir3_column] = 1

    # Get rain probability
    probability = loaded_model.predict_proba(input_data)[0][1]

    # Apply our selected threshold
    prediction = "Yes" if probability >= loaded_threshold else "No"

    return prediction, probability
```
